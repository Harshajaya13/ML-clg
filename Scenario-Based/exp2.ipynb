```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.xls")
df.head()
df.tail()

# 2. Count missing values in each column
df.isnull().sum()

# Check data types
print(df.dtypes, end="\n\n")

# find the missing values
print(df.isnull().sum())

# fill the missing values 
df['Age'] = df['Age'].fillna(df['Age'].mean())

# so we need to use the mode to fill the cabin and embarked values
embarked_mode = df['Embarked'].mode()[0]
df['Embarked'] = df['Embarked'].fillna(embarked_mode)

# fill the missing values of the cabin with the mode
cabin_mode = df['Cabin'].mode()[0]
df['Cabin'] = df['Cabin'].fillna(cabin_mode)

print(end="\n\n")

# find if any leftovers 
print(df.isnull().sum())

# one hot encoding
df = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True,dtype=int)
df
```
