```python
import numpy as np
import pandas as pd
import statistics as ss

df = pd.read_csv("diabetes.csv")
df.head()
df.tail()
# print the datatypes of the columns
print(df.dtypes,end="\n\n\n\n\n")

# print to check whether any null values exists
print(df.isnull().sum(),end="\n\n\n\n")

# to print the numerical data data we use the 
print(df.select_dtypes(include=['number']).columns)
# replace the number with the object for the categorical columns

# dropping the columns we can use this 
df.drop(columns=['Pregnancies'],inplace=True)
df.head()
```

```python
mean = np.mean(df['Glucose'])
mean

variance = np.var(df['Glucose'])
variance

std  = np.std(df['Glucose'])
std

mode = ss.mode(df['Glucose'])
mode

maximum = np.max(df['Glucose'])
maximum

minimum = np.min(df['Glucose'])
minimum

Q1  = df['Glucose'].quantile(0.25)
Q3 = df['Glucose'].quantile(0.75)
Q1
Q3

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Glucose'] < lower_bound) | (df['Glucose'] > upper_bound)]
print(outliers)
```
